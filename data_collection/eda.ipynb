```python
import pandas as pd
import numpy as np 

import matplotlib.pyplot as plt
import seaborn as sns

pd.set_option("display.max_columns",None)
pd.set_option("display.max_rows",None)
sns.set_theme(
    style="whitegrid"
)
```

```python
df = pd.read_excel("../izto_tum_firmalar_temiz.xlsx", dtype=str)
print("veri seti özeti")
df.info()
df.head(10)
```

```python
df.nunique()
```

```python
k_df=pd.DataFrame(
    {
        "dolu satır sayısı" : df.count(),
        "benzersiz değer sayısı" : df.nunique(),
        "tekrar eden veri sayısı" : df.count() - df.nunique()
    }
)

print("veri seti benzersiz değer sayısı ve tekrar eden veri sayısı")
k_df
```

```python
df.duplicated().sum()
```

```python
unvan_kopyaları = df[df.duplicated(subset=["Ünvanı"],keep=False)].sort_values(by="Ünvanı")
unvan_kopyaları
```

```python
sicil_kopyaları = df[
    df.duplicated(
        subset=["Ticari Sicil No"],
        keep=False
    )
].sort_values(
    by="Ticari Sicil No"
)
display(sicil_kopyaları.
    head(10)
)
```

```python
df["İlçe"]=df["İlçe"].fillna("Bilinmiyor")
kopya_sayısı = df.duplicated(
    subset=["Ünvanı","Ticari Sicil No"],
    keep=False
    ).sum()
if kopya_sayısı > 0:
    df = df.drop_duplicates(subset=["Ünvanı", "Ticari Sicil No"])
print(f"Ünvan ve Ticari Sicil No alanlarında {kopya_sayısı} adet kopya veri bulunmaktadır.")
```

```python
plt.figure(figsize=(8,5))
sns.countplot(
    data=df,
    y="İlçe",
    order=df["İlçe"].value_counts().index[:10],
    palette="viridis"
)
plt.title("izmir - en çok firma bulunan 10 ilçe")
plt.xlabel("firma sayısı")
plt.ylabel("ilçe")  
plt.show
```

```python
df["dijital_varlıklar"]=df["Web Adresi"].notna().map( #notna() → boş mu dolu mu kontrol eder. #
    {
        True: "web sitesi var",
        False:"web sitesi yok"
        }

)
plt.figure(figsize=(8,5))
df["dijital_varlıklar"].value_counts().plot.pie(
    autopct="%1.1f%%"
)

plt.title("firma web sitesi var mı yok mu")
plt.show()
```

```python
# 1. Pandas'ın otomatik numaralandırma (factorize) özelliğini kullanarak doğrudan ID sütunları oluşturalım
df['ilce_id'] = pd.factorize(df['İlçe'])[0] + 1
df['meslek_id'] = pd.factorize(df['Meslek Grubu'])[0] + 1

# 2. ID ve metin yan yanayken bunları kopyalayıp, kopyaları silelim (SQL'deki SELECT DISTINCT mantığı)
df_ilceler = df[['ilce_id', 'İlçe']].drop_duplicates().rename(columns={'İlçe': 'ilce_adi'})
df_meslekler = df[['meslek_id', 'Meslek Grubu']].drop_duplicates().rename(columns={'Meslek Grubu': 'meslek_adi'})

# 3. Eski uzun metin sütunlarını ana tablodan silelim
df_firmalar = df.drop(columns=['İlçe', 'Meslek Grubu'])

# Tabloların son haline göz atalım
display(df_ilceler.head(3))
display(df_firmalar.head(3))
```

```python

from sqlalchemy import create_engine

# MySQL bağlantı motorunu oluşturuyoruz (Veritabanının kapısını çalıyoruz)
sifre = "change35" # Buraya kendi şifreni yaz
engine = create_engine(f"mysql+pymysql://root:{sifre}@127.0.0.1:3306/staj_db")

print("Veritabanına bağlanılıyor...")

# 1. İlçeler tablosunu aktarıyoruz
df_ilceler.to_sql(name='ilceler', con=engine, if_exists='replace', index=False)
print(" 'ilceler' tablosu MySQL'e başarıyla aktarıldı!")

# 2. Meslek Grupları tablosunu aktarıyoruz
df_meslekler.to_sql(name='meslek_gruplari', con=engine, if_exists='replace', index=False)
print(" 'meslek_gruplari' tablosu MySQL'e başarıyla aktarıldı!")

# 3. Güncellenmiş Ana Tabloyu aktarıyoruz
df_firmalar.to_sql(name='firmalar', con=engine, if_exists='replace', index=False)
print(" 'firmalar' ana tablosu MySQL'e başarıyla aktarıldı!")
```
